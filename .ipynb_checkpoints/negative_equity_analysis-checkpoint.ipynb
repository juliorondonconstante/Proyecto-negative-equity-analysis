# 🏠 Negative Equity en el Mercado Inmobiliario de EE.UU.
## Análisis de datos 2011 - 2017

---

### ¿Cuál es el problema?

Imagina que compraste una casa por 200,000 dólares con una hipoteca. 
Dos años después, por la crisis económica, tu casa ahora vale 
150,000 dólares — pero todavía debes 180,000 dólares al banco.

Eso es el **negative equity** — debes más de lo que vale tu casa.
Durante la crisis financiera del 2008, millones de familias 
estadounidenses quedaron atrapadas en esta situación.

---

### ¿Qué vamos a analizar?

Con datos de **Zillow Research** analizamos:
- ¿Cuándo fue el peor momento a nivel nacional?
- ¿Qué estados y ciudades sufrieron más?
- ¿Cómo fue la recuperación entre 2012 y 2017?
- ¿Hubo patrones geográficos en la crisis?

---

| Trimestres analizados | 24 |

## 2. ¿Cuándo fue el peor momento?

Antes de mirar estados o ciudades, entendamos el panorama completo 
del país. ¿Cuándo alcanzó la crisis su punto máximo y qué tan rápido 
se recuperó?

### Lo que esperamos encontrar

La crisis financiera del 2008 golpeó el mercado inmobiliario con 
fuerza. Los precios de las casas colapsaron entre 2008 y 2012, 
por lo que esperamos ver el pico del negative equity alrededor 
de ese período.

### Lo que encontramos

```python
# Extraemos los valores de EE.UU. para cada trimestre
valores_nacional = [nacional[q] for q in quartes]

# Graficamos
plt.figure(figsize=(12, 5))
plt.plot(quartes, valores_nacional, color='steelblue', linewidth=2.5)

plt.title('Negative Equity en EE.UU. — 2011 a 2017')
plt.ylabel('% de viviendas')
plt.xlabel('Trimestre')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Interpretación

- El pico nacional fue **31.4% en 2012Q1** — casi 1 de cada 3 
  hogares debía más de lo que valía su casa
- A partir de 2012 la recuperación fue constante y sostenida 
  durante 5 años consecutivos
- Para **2017Q1 el índice bajó a 10.4%** — una reducción del 67% 
  desde el pico

> 📌 La recuperación no fue un evento sino un proceso gradual 
> de 5 años. Esto es importante para entender los análisis 
> regionales que vienen a continuación.

```python
valores_nacional = [nacional[q].iloc[0] for q in quartes]

# Encontramos el pico automaticamente
pico_valor  = max(valores_nacional)
pico_indice = valores_nacional.index(pico_valor)

plt.figure(figsize=(12, 5))
plt.plot(quartes, valores_nacional, color='steelblue', linewidth=2.5)

# Marcamos el pico en rojo
plt.scatter(pico_indice, pico_valor, color='red', s=80, zorder=5)
plt.annotate(
    f'Pico: {pico_valor:.1%} ({quartes[pico_indice]})',
    xy=(pico_indice, pico_valor),
    xytext=(pico_indice + 1, pico_valor + 0.01),
    fontsize=9, color='red',
    arrowprops=dict(arrowstyle='->', color='red')
)

# Marcamos el valor final en verde
ultimo_indice = len(quartes) - 1
plt.scatter(ultimo_indice, valores_nacional[-1], color='green', s=80, zorder=5)
plt.annotate(
    f'2017Q1: {valores_nacional[-1]:.1%}',
    xy=(ultimo_indice, valores_nacional[-1]),
    xytext=(ultimo_indice - 5, valores_nacional[-1] + 0.01),
    fontsize=9, color='green',
    arrowprops=dict(arrowstyle='->', color='green')
)

plt.title('Negative Equity en EE.UU. — 2011 a 2017')
plt.ylabel('% de viviendas')
plt.xlabel('Trimestre')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Resumen en texto
caida = pico_valor - valores_nacional[-1]
print(f"Pico            : {pico_valor:.1%} en {quartes[pico_indice]}")
print(f"Valor en 2017Q1 : {valores_nacional[-1]:.1%}")
print(f"Reducción total : {caida:.1%}")
```

## 3. ¿Qué estados sufrieron más?

El promedio nacional esconde realidades muy distintas entre estados. 
Un 31% nacional puede significar que todos los estados estaban en 31% 
— o que algunos estaban en 70% y otros en 10%.

Vamos a descubrir cuál fue el caso.

### Lo que esperamos encontrar

Los estados del "Sun Belt" — la franja sur del país que vivió un 
boom inmobiliario antes de la crisis — deberían aparecer como los 
más afectados. Nevada, Arizona y Florida fueron los epicentros 
del boom de construcción pre-crisis.

```python
# Calculamos el pico de cada estado (el valor máximo que tuvo en cualquier trimestre)
estados['pico'] = estados[quartes].max(axis=1)

# Valor en 2017Q1
estados['valor_2017'] = estados['2017Q1']

# Cuánto bajó desde el pico hasta 2017
estados['recuperacion'] = ((estados['pico'] - estados['valor_2017']) / estados['pico'] * 100).round(1)

# Top 10 estados más afectados
top10 = estados.nlargest(10, 'pico')[['RegionName', 'pico', 'valor_2017', 'recuperacion']]

print(top10.to_string(index=False))
```

```python
plt.figure(figsize=(10, 6))

# Barras del pico en rojo
plt.barh(top10['RegionName'], top10['pico'], color='red', alpha=0.6, label='Pico histórico')

# Barras del valor 2017 en azul encima
plt.barh(top10['RegionName'], top10['valor_2017'], color='steelblue', alpha=0.9, label='2017Q1')

# Etiqueta de recuperación al lado de cada barra
for i, (_, row) in enumerate(top10.iterrows()):
    plt.text(
        row['pico'] + 0.005,
        i,
        f"{row['recuperacion']}% ↓",
        va='center',
        fontsize=9,
        color='green'
    )

plt.xlabel('% de viviendas con negative equity')
plt.title('Top 10 Estados más afectados — Pico vs 2017Q1')
plt.legend()
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### Lo que encontramos en el ranking

- **Nevada (66.8%)** fue el caso más extremo — 2 de cada 3 casas 
  estaban bajo el agua en su peor momento
- **Arizona y Georgia** le siguen con más del 46%
- **California** tuvo el pico más alto del top 10 pero también 
  la recuperación más rápida con 82.7%
- **Illinois** fue el que menos se recuperó con solo 57.8% — 
  todavía tenía mucho negative equity en 2017

```python
# Preparamos los datos: estados como filas, trimestres como columnas
heat_data = estados.set_index('RegionName')[quartes]

plt.figure(figsize=(16, 10))
sns.heatmap(
    heat_data,
    cmap='RdYlGn_r',        # rojo = alto negative equity, verde = bajo
    linewidths=0.3,
    linecolor='white',
    cbar_kws={'label': '% negative equity'}
)

plt.title('Negative Equity por Estado y Trimestre — 2011 a 2017')
plt.xlabel('')
plt.ylabel('')
plt.xticks(rotation=45, ha='right', fontsize=8)
plt.yticks(rotation=0, fontsize=8)
plt.tight_layout()
plt.show()
```

### Lo que nos dice el mapa de calor

El heatmap confirma visualmente lo del ranking — todos los estados 
van de colores cálidos en 2011 hacia verde en 2017. Pero hay dos 
patrones importantes:

- **Nevada** resalta solo en rojo intenso — su crisis fue tan grave 
  que se separa visualmente de todos los demás estados
- **Los blancos** son datos faltantes — estados pequeños donde 
  Zillow no tenía suficientes transacciones en los primeros trimestres

> 📌 El heatmap permite ver 48 estados y 24 trimestres en una sola 
> imagen — algo imposible de lograr con una tabla de números.

## 4. ¿Hubo patrones de recuperación?

Sabemos qué estados sufrieron más. Pero ¿se recuperaron todos 
de la misma manera? ¿Hubo grupos de estados con comportamientos 
similares?

Para responder esto usamos **KMeans** — un algoritmo de machine 
learning que agrupa automáticamente los estados según cómo 
evolucionó su negative equity a lo largo del tiempo.

### ¿Cómo funciona KMeans?

Imagina que tienes 48 líneas en una gráfica — una por estado. 
KMeans las agrupa según su forma:

- Las que empezaron alto y bajaron lento → un grupo
- Las que empezaron alto y bajaron rápido → otro grupo
- Las que siempre estuvieron bajas → otro grupo

El algoritmo hace esto matemáticamente sin que nosotros 
le digamos cuáles son los grupos — los descubre solo.

### ¿Cuántos grupos usar?

Usamos el **método del codo** para determinarlo — probamos 
con 1 hasta 10 grupos y elegimos donde la mejora empieza 
a ser mínima.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Usamos solo estados con datos completos
# Extraemos solo los valores numéricos de los trimestres
X = estados_limpios[quartes].values

# Normalizamos los datos para que el algoritmo funcione bien
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Aplicamos KMeans con 4 grupos
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
estados_limpios = estados_limpios.copy()
estados_limpios['cluster'] = kmeans.fit_predict(X_scaled)

# Vemos qué estados quedaron en cada grupo
for cluster in sorted(estados_limpios['cluster'].unique()):
    estados_grupo = estados_limpios[estados_limpios['cluster'] == cluster]['RegionName'].tolist()
    print(f"\nGrupo {cluster}:")
    print(', '.join(estados_grupo))
```

```python
# ✅ Definimos primero las listas antes de usarlas
colores   = ['steelblue', 'red', 'orange', 'green']
etiquetas = ['Grupo 0 - Crisis baja', 'Grupo 1 - Crisis severa',
             'Grupo 2 - Crisis moderada', 'Grupo 3 - Recuperación rápida']

plt.figure(figsize=(12, 6))

for cluster in sorted(estados_limpios['cluster'].unique()):
    grupo    = estados_limpios[estados_limpios['cluster'] == cluster]
    promedio = grupo[quartes].mean()

    plt.plot(
        range(len(quartes)),
        promedio.values,
        color=colores[cluster],
        linewidth=2.5,
        label=etiquetas[cluster]
    )

posiciones  = list(range(0, len(quartes), 4))
etiquetas_x = [quartes[i][:4] for i in posiciones]
plt.xticks(posiciones, etiquetas_x)

plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.0%}'))
plt.title('Patrones de recuperación por grupo de estados')
plt.ylabel('% de viviendas con negative equity')
plt.xlabel('Año')
plt.legend()
plt.tight_layout()
plt.show()
```

### Los 4 patrones que encontró el algoritmo

| Grupo | Estados representativos | Característica |
|---|---|---|
| Crisis baja | New York, Alaska, Montana | Nunca superaron el 20% |
| Crisis moderada | Ohio, Virginia, Indiana | Centro del país, recuperación constante |
| Crisis severa | Nevada, Florida, Michigan | Epicentro de la crisis, recuperación lenta |
| Recuperación rápida | California, Washington, Colorado | Crisis alta pero bajaron más rápido que todos |

> 📌 El algoritmo confirmó matemáticamente lo que vimos 
> visualmente en el heatmap — Nevada y sus vecinos forman 
> un grupo completamente distinto al resto del país.

```python
# Probamos con 1 hasta 10 clusters y medimos el error de cada uno
inercias = []
rango = range(1, 11)

for k in rango:
    modelo = KMeans(n_clusters=k, random_state=42, n_init=10)
    modelo.fit(X_scaled)
    inercias.append(modelo.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(rango, inercias, marker='o', color='steelblue', linewidth=2)
plt.axvline(x=4, color='red', linestyle='--', label='Nuestro k=4')
plt.title('Método del Codo — ¿Cuántos clusters usar?')
plt.xlabel('Número de clusters (k)')
plt.ylabel('Inercia')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, estados_limpios['cluster'])
print(f"Silhouette Score : {score:.3f}")
print()
if score > 0.5:
    print("✅ Clusters bien definidos y separados")
elif score > 0.25:
    print("⚠️ Clusters razonables pero con algo de solapamiento")
else:
    print("❌ Los clusters no están bien definidos")
```

### Evaluación del modelo

| Métrica | Resultado | Interpretación |
|---|---|---|
| Método del codo | k=4 en la zona correcta | ✅ Número adecuado de grupos |
| Silhouette Score | 0.330 | ⚠️ Agrupación aceptable |

> 💡 Un Silhouette Score de 0.330 es normal para datos geográficos 
> — los estados no tienen fronteras perfectas entre sí, siempre 
> habrá algo de solapamiento entre grupos vecinos.

```python
from sklearn.decomposition import PCA

# Reducimos a 2 dimensiones para poder graficar
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 6))

for cluster in sorted(estados_limpios['cluster'].unique()):
    mascara = estados_limpios['cluster'].values == cluster
    plt.scatter(
        X_pca[mascara, 0],
        X_pca[mascara, 1],
        c=colores[cluster],
        label=etiquetas[cluster],
        s=80,
        alpha=0.85,
        edgecolors='white',
        linewidth=0.5
    )

# Anotamos algunos estados clave
estados_anotar = ['Nevada', 'Florida', 'California', 'Illinois', 'New York']
for i, (_, row) in enumerate(estados_limpios.iterrows()):
    if row['RegionName'] in estados_anotar:
        plt.annotate(
            row['RegionName'],
            (X_pca[i, 0], X_pca[i, 1]),
            fontsize=8,
            xytext=(5, 5),
            textcoords='offset points'
        )

varianza = pca.explained_variance_ratio_
plt.title(f'Visualización PCA de Clusters\nVarianza explicada: PC1={varianza[0]:.0%}, PC2={varianza[1]:.0%}')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Calculamos el pico de cada metro
metros_limpios = metros_limpios.copy()
metros_limpios['pico'] = metros_limpios[quartes].max(axis=1)

# Valor en 2017Q1
metros_limpios['valor_2017'] = metros_limpios['2017Q1']

# Cuánto se recuperó
metros_limpios['recuperacion'] = ((metros_limpios['pico'] - metros_limpios['valor_2017']) / metros_limpios['pico'] * 100).round(1)

# Top 10
top10_metros = metros_limpios.nlargest(10, 'pico')[['RegionName', 'StateName', 'pico', 'valor_2017', 'recuperacion']]

print(top10_metros.to_string(index=False))
```

### ¿Qué nos dice el PCA?

El PCA comprime los 24 trimestres en 2 dimensiones para 
poder visualizar los grupos. Con solo 2 dimensiones capturamos 
el **97% de la información** del dataset — casi nada se pierde.

Lo más revelador de esta gráfica:

- **Nevada** aparece completamente solo en el extremo derecho 
  — ningún estado se comportó como él
- **El solapamiento entre azul y naranja** en el centro confirma 
  el Silhouette Score de 0.330 — esos estados son difíciles 
  de clasificar incluso para el algoritmo
- **California** queda separado verticalmente — su velocidad 
  de recuperación fue tan única que lo aleja de su propio grupo

## 5. ¿Qué pasó dentro de los estados más golpeados?

Hasta ahora analizamos el país completo y los 48 estados. 
Pero un estado es muy grande — puede esconder realidades 
muy distintas entre sus ciudades.

Por ejemplo: ¿fue igual de grave la crisis en Las Vegas 
que en Reno? ¿O hubo zonas dentro de Nevada muchísimo 
peores que otras?

Para responder esto bajamos al nivel de **áreas metropolitanas** 
— las 796 zonas urbanas del dataset.

- **Pero primero revisamos los datos de los metros**:



```python
# Contamos cuántos trimestres tiene cada metro con dato
metros['trimestres_completos'] = metros[quartes].notna().sum(axis=1)

print("Distribución de trimestres completos por metro:")
print(metros['trimestres_completos'].value_counts().sort_index(ascending=False).head(10))
print(f"\nTotal trimestres posibles: {len(quartes)}")
```

```python
# Filtramos metros con al menos 20 trimestres completos
metros_limpios = metros[metros['trimestres_completos'] >= 20].copy()

print(f"Metros originales          : {len(metros)}")
print(f"Metros eliminados          : {len(metros) - len(metros_limpios)}")
print(f"Metros con datos confiables: {len(metros_limpios)}")
print(f"Porcentaje conservado      : {len(metros_limpios)/len(metros)*100:.1f}%")

# Recalculamos las métricas con los metros limpios
metros_limpios['pico']         = metros_limpios[quartes].max(axis=1)
metros_limpios['valor_2017']   = metros_limpios['2017Q1']
metros_limpios['recuperacion'] = ((metros_limpios['pico'] - metros_limpios['valor_2017']) / metros_limpios['pico'] * 100).round(1)

print("\n✓ Métricas recalculadas con datos confiables")
```

```python
top10_metros = metros_limpios.nlargest(10, 'pico')[
    ['RegionName', 'StateName', 'pico', 'valor_2017', 'recuperacion', 'trimestres_completos']
]

print(top10_metros.to_string(index=False))
```

### Decisión metodológica

De los 796 metros originales, el 38% tenía valores nulos 
en algunos trimestres. Aplicamos un filtro estricto:

- Exigimos mínimo **20 de 24 trimestres completos**
- Conservamos **681 metros (85,6% del total)**
- El top 10 no varió con este criterio — validando la robustez 
  del análisis

```python
plt.figure(figsize=(11, 6))

# Barras del pico en rojo
plt.barh(top10_metros['RegionName'], top10_metros['pico'],
         color='red', alpha=0.6, label='Pico histórico')

# Barras del valor 2017 en azul
plt.barh(top10_metros['RegionName'], top10_metros['valor_2017'],
         color='steelblue', alpha=0.9, label='2017Q1')

# Etiqueta de recuperación
for i, (_, row) in enumerate(top10_metros.iterrows()):
    plt.text(
        row['pico'] + 0.005,
        i,
        f"{row['recuperacion']}% ↓",
        va='center',
        fontsize=9,
        color='green'
    )

# Coloreamos el nombre según el estado
colores_estado = {'NV': 'red', 'CA': 'steelblue', 'MN': 'green'}
ax = plt.gca()
for label in ax.get_yticklabels():
    nombre = label.get_text()
    fila = top10_metros[top10_metros['RegionName'] == nombre]
    if not fila.empty:
        estado = fila['StateName'].values[0]
        label.set_color(colores_estado.get(estado, 'black'))

plt.xlabel('% de viviendas con negative equity')
plt.title('Top 10 Metros más afectados — Pico vs 2017Q1\n(Rojo=NV, Azul=CA, Verde=MN)')
plt.legend()
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### Lo que encontramos

- **Fernley, NV (71.8%)** fue el metro más afectado de todo 
  el país — incluso peor que el promedio de Nevada como estado
- **4 de los 10 metros son de Nevada** — confirma que el estado 
  era el epicentro absoluto de la crisis
- **5 metros son de California** — aunque California se recuperó 
  bien como estado, sus ciudades del interior contaban 
  una historia muy distinta
- **Owatonna, MN** es la gran sorpresa — Minnesota no aparecía 
  como estado crítico pero este metro específico tuvo un 
  pico de 66%

> 📌 Este es el valor del zoom — los promedios estatales 
> esconden realidades locales muy distintas. Un analista 
> que solo mira estados se pierde la historia real.

```python
# Filtramos metros de los 3 estados más golpeados
estados_criticos = ['NV', 'AZ', 'FL']
metros_criticos  = metros_limpios[metros_limpios['StateName'].isin(estados_criticos)]

# Graficamos la trayectoria de cada metro
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

colores_estado = {'NV': 'red', 'AZ': 'orange', 'FL': 'steelblue'}

for ax, estado in zip(axes, estados_criticos):
    grupo = metros_criticos[metros_criticos['StateName'] == estado]
    
    for _, row in grupo.iterrows():
        valores = [row[q] for q in quartes if q in row.index]
        ax.plot(
            range(len(quartes)),
            valores,
            color=colores_estado[estado],
            alpha=0.5,
            linewidth=1.2
        )
    
    # Promedio del estado
    promedio = grupo[quartes].mean()
    ax.plot(
        range(len(quartes)),
        promedio.values,
        color='black',
        linewidth=2.5,
        linestyle='--',
        label='Promedio'
    )
    
    posiciones  = list(range(0, len(quartes), 4))
    etiquetas_x = [quartes[i][:4] for i in posiciones]
    ax.set_xticks(posiciones)
    ax.set_xticklabels(etiquetas_x)
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.0%}'))
    ax.set_title(f'{estado} — {len(grupo)} metros')
    ax.legend(fontsize=8)

fig.suptitle('Trayectoria de metros dentro de los estados más afectados\n(línea negra = promedio del estado)',
             fontsize=12)
plt.tight_layout()
plt.show()
```

### ¿Fue la crisis pareja dentro de cada estado?

| Estado | Metros analizados | ¿Crisis pareja? |
|---|---|---|
| Nevada | 8 | ❌ Muy desigual — Fernley fue caso extremo |
| Arizona | 10 | ✅ Homogénea — todos se comportaron similar |
| Florida | 28 | ✅ Homogénea pero recuperación más lenta |

> 📌 Nevada no solo fue el estado más golpeado — también 
> fue el más desigual internamente. Mientras Fernley llegó 
> al 71.8%, otros metros de Nevada nunca superaron el 40%.

### Mapa 1 — El pico histórico por estado

¿Dónde fue más grave la crisis? Los estados más rojos son

negative equity en su peor momento.

```python
import plotly.express as px

# Preparamos los datos
mapa_data = estados[['StateName', 'pico', 'valor_2017', 'recuperacion']].copy()
mapa_data = mapa_data.dropna(subset=['StateName', 'pico'])

# Creamos el mapa
fig = px.choropleth(
    mapa_data,
    locations='StateName',          # columna con las siglas de los estados
    locationmode='USA-states',      # modo de ubicación por estado
    color='pico',                   # columna que define el color
    scope='usa',                    # solo mostramos EE.UU.
    color_continuous_scale='RdYlGn_r',  # rojo = peor, verde = mejor
    hover_name='StateName',         # qué aparece al pasar el mouse
    hover_data={
        'pico': ':.1%',
        'valor_2017': ':.1%',
        'recuperacion': ':.1f'
    },
    labels={
        'pico': 'Pico histórico',
        'valor_2017': 'Valor 2017Q1',
        'recuperacion': '% Recuperado'
    },
    title='Negative Equity por Estado — Pico Histórico (2011-2017)'
)

fig.update_layout(
    coloraxis_colorbar=dict(
        title='% Negative Equity',
        tickformat='.0%'
    )
)

fig.show()

fig.write_html('mapa_negative_equity.html')
print("✓ Mapa guardado como mapa_negative_equity.html")
```

## 7. Conclusiones

### ¿Qué aprendimos?

El negative equity en EE.UU. entre 2011 y 2017 no fue un 
problema uniforme — fue una crisis con geografía, con ganadores 
y perdedores muy distintos dependiendo de dónde vivías.

### Hallazgos principales

| Hallazgo | Detalle |
|---|---|
| Pico nacional | 31.4% en 2012Q1 — 1 de cada 3 hogares |
| Recuperación nacional | Bajó a 10.4% en 2017 — reducción del 67% |
| Estado más golpeado | Nevada con 66.8% en su pico |
| Metro más golpeado | Fernley, NV con 71.8% |
| Recuperación más rápida | California con 82.7% desde el pico |
| Recuperación más lenta | Illinois con solo 57.8% |

### Los 3 aprendizajes más importantes

**1. La geografía importa más que el promedio**
El promedio nacional de 31.4% no le dice nada a un propietario 
de Nevada con 66.8% ni a uno de New York con 15%. Analizar 
solo promedios nacionales esconde las historias reales.

**2. La velocidad de recuperación fue tan importante como la crisis**
California y Nevada tuvieron crisis similares pero California 
se recuperó casi el doble de rápido. La fortaleza económica 
de base de cada región determinó su capacidad de rebote.

**3. El zoom geográfico revela lo que el promedio esconde**
Nevada como estado tenía 66.8% de pico. Pero Fernley dentro 
de Nevada llegó al 71.8% mientras otros metros del mismo 
estado no superaban el 40%. Sin el análisis de metros 
esa realidad sería invisible.
